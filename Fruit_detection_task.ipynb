```python
import yaml
import os
data_yaml_path = '/content/data.yaml'

correct_dataset_root_absolute = '/content/fruit-detection-dataset/fruit-detection-dataset'

with open(data_yaml_path, 'r') as f:
    data_config = yaml.safe_load(f)
data_config['path'] = correct_dataset_root_absolute

with open(data_yaml_path, 'w') as f:
    yaml.safe_dump(data_config, f)

model = YOLO('yolo11m.pt')

results = model.train(
    data=data_yaml_path,
    epochs=50,
    imgsz=640,
    batch=-1,
    patience=10,
    name='fruit_detection_mentor_finetuning'
)
```

```python
best_model = YOLO('/content/runs/detect/fruit_detection_mentor_finetuning-2/weights/best.pt')

results = best_model.predict(source='/content/bunches-of-yellow-and-green-bananas-honduras.webp', save=True, conf=0.01)
```
